```python
ff = []
mm = []
mf = []
for i, j in G.edges:
    sens_i = i in data['communities'][0]
    sens_j = j in data['communities'][0]
    
    if sens_i and sens_j:
        ff.append((i, j))
        
    if sens_i ^ sens_j:
        mf.append((i, j))
        
    if not sens_i and not sens_j:
        mm.append((i, j))
```
